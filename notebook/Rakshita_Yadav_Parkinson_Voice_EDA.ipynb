```python
# ==============================================================================
# Project Title: Parkinson Voice Measurements EDA
# Name: Rakshita Yadav | Roll Number: 2026B01010141
# Class / Section: CSE 11 | Course / Department: B.Tech CSE
# Faculty Guide: Faizan Khawar
# Requirement: Import Python, NumPy, Pandas, and load the UCI Parkinsons dataset.
# ==============================================================================

!pip install -q ucimlrepo

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from ucimlrepo import fetch_ucirepo

# Fetch dataset (ID: 174) and combine features and status target into one DataFrame
parkinsons = fetch_ucirepo(id=174)
df = pd.concat([parkinsons.data.features, parkinsons.data.targets], axis=1)

print("Dataset loaded successfully!")
```

```python
# Requirement: Display head(), tail(), shape, info(), and describe().

print("1. First 5 rows (head):")
display(df.head())

print("\n2. Last 5 rows (tail):")
display(df.tail())

print("\n3. Dataset Shape (Rows, Columns):", df.shape)

print("\n4. Dataset Info:")
df.info()

print("\n5. Statistical Summary (describe):")
display(df.describe())
```

```python
# Question 1: How many records are in each status group?

print("Record count in each status group (0 = Healthy, 1 = Parkinson's):")
print(df['status'].value_counts())

print("\nPercentage breakdown:")
print((df['status'].value_counts(normalize=True) * 100).round(2))
```

```python
# Question 2: Calculate mean, median, minimum and maximum for selected voice measurements
# such as MDVP:Fo(Hz), MDVP:Jitter(%) and MDVP:Shimmer (using NumPy arrays).

# Column indices: 0 = MDVP:Fo(Hz), 3 = MDVP:Jitter(%), 8 = MDVP:Shimmer
selected_indices = [0, 3, 8]
labels = ['MDVP:Fo(Hz)', 'MDVP:Jitter(%)', 'MDVP:Shimmer']

for idx, label in zip(selected_indices, labels):
    arr = df.iloc[:, idx].to_numpy()
    print(f"Measurement: {label}")
    print(f"  Mean   : {np.mean(arr):.4f}")
    print(f"  Median : {np.median(arr):.4f}")
    print(f"  Minimum: {np.min(arr):.4f}")
    print(f"  Maximum: {np.max(arr):.4f}\n")
```

```python
# Question 3: Compare the average of these measurements by status.

subset = df.iloc[:, [0, 3, 8]].copy()
subset.columns = ['MDVP:Fo(Hz)', 'MDVP:Jitter(%)', 'MDVP:Shimmer']
subset['status'] = df['status']

print("Average of selected voice measurements grouped by status:")
display(subset.groupby('status').mean())
```

```python
# Question 4: Find the 10 highest values for one selected measurement (MDVP:Fo(Hz)).
# Also demonstrates Pandas filtering and sorting.

# Pandas Filtering: Filter records where status == 1 and MDVP:Fo > 200
high_fo_filtered = df[(df['status'] == 1) & (df['MDVP:Fo'] > 200)]
print(f"Filtered count (status == 1 and MDVP:Fo > 200 Hz): {len(high_fo_filtered)} records\n")

# Pandas Sorting: Top 10 highest values for MDVP:Fo(Hz)
print("Top 10 highest values for MDVP:Fo(Hz):")
top_10 = df[['MDVP:Fo', 'MDVP:Fhi', 'MDVP:Flo', 'status']].sort_values(
    by='MDVP:Fo', ascending=False
).head(10)

display(top_10)
```

```python
# Question 5: Check for missing values and duplicates.

print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values in dataset:", df.isnull().sum().sum())
print("Total duplicate rows in dataset :", df.duplicated().sum())
```

```python
# Requirement: Make a status-count chart and a histogram.

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Chart 1: Status-Count Bar Chart
counts = df['status'].value_counts().sort_index()
ax1.bar(['Status 0', 'Status 1'], counts.values, color=['#4C72B0', '#DD8452'], edgecolor='black', width=0.5)
ax1.set_title('Record Count by Status Group', fontweight='bold')
ax1.set_xlabel('Status Category')
ax1.set_ylabel('Number of Recordings')
for i, v in enumerate(counts.values):
    ax1.text(i, v + 3, str(v), ha='center', fontweight='bold')

# Chart 2: Histogram of MDVP:Fo(Hz)
ax2.hist(df['MDVP:Fo'], bins=15, color='#55A868', edgecolor='black')
ax2.set_title('Distribution of MDVP:Fo(Hz)', fontweight='bold')
ax2.set_xlabel('MDVP:Fo(Hz)')
ax2.set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
